```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# ── Load data ─────────────────────────────────────────────────────────────────
# Replace with your actual file path
dfs = ["cannabis.csv", "cocaine.csv", "hashish.csv"]
for df_name in dfs:
    print(f"\n=== Analyzing {df_name} ===")
    df = pd.read_csv(rf"C:\Users\user\OneDrive\טל\projects\justice-tal\bias\data\impute\{df_name}")

    # Keep only cases with actual imprisonment (punishment_months > 0)
    # and with declared sentencing bounds
    df = df.dropna(subset=["min_punishment", "max_punishment", "punishment_months"])
    df = df[df["punishment_months"] > 0]

    print(f"N = {len(df)} cases with complete data\n")

    # ── Correlations ──────────────────────────────────────────────────────────────
    for bound, label in [("min_punishment", "Lower Bound"), ("max_punishment", "Upper Bound")]:
        pearson_r, pearson_p = stats.pearsonr(df[bound], df["punishment_months"])
        spearman_r, spearman_p = stats.spearmanr(df[bound], df["punishment_months"])
        print(f"{label} → Actual Sentence:")
        print(f"  Pearson  r = {pearson_r:.3f}, p = {pearson_p:.2e}")
        print(f"  Spearman r = {spearman_r:.3f}, p = {spearman_p:.2e}")
        print()

    # ── OLS: how much variance in actual sentence is explained by bounds? ─────────
    from sklearn.linear_model import LinearRegression
    from sklearn.metrics import r2_score

    X = df[["min_punishment", "max_punishment"]].values
    y = df["punishment_months"].values

    model = LinearRegression().fit(X, y)
    r2 = r2_score(y, model.predict(X))
    print(f"R² (both bounds → actual sentence): {r2:.3f}")
    print(f"  β_lower = {model.coef_[0]:.3f}")
    print(f"  β_upper = {model.coef_[1]:.3f}")
    print(f"  intercept = {model.intercept_:.3f}\n")

    # ── Plot ──────────────────────────────────────────────────────────────────────
    fig, axes = plt.subplots(1, 2, figsize=(10, 4.5))

    for ax, (bound, label) in zip(axes, [
        ("min_punishment", "Lower Bound (months)"),
        ("max_punishment", "Upper Bound (months)")
    ]):
        r, p = stats.spearmanr(df[bound], df["punishment_months"])

        ax.scatter(df[bound], df["punishment_months"],
                alpha=0.3, s=20, color="#2E75B6", edgecolors="none")

        # Regression line
        m, b = np.polyfit(df[bound], df["punishment_months"], 1)
        x_line = np.linspace(df[bound].min(), df[bound].max(), 100)
        ax.plot(x_line, m * x_line + b, color="#C00000", linewidth=1.5)

        ax.set_xlabel(label, fontsize=11)
        ax.set_ylabel("Actual Sentence Length (months)", fontsize=11)
        ax.set_title(f"{label.split(' (')[0]} vs. Actual Sentence\n"
                    f"Spearman r = {r:.2f}, p < .001", fontsize=11)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

    plt.tight_layout()
    # plt.savefig(rf"plots\{df_name}_bounds_correlation.png", dpi=150, bbox_inches="tight")
    plt.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# ── הגדרות גופן גלובליות (צר, גבוה ומודגש) ───────────────
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial Narrow', 'Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['font.stretch'] = 'condensed' 
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titleweight'] = 'bold'

# ── Load data ──────────────────────────────────────────────────────
base_path = r'C:\Users\user\OneDrive\טל\projects\justice-tal\bias\matched_outputs\impute'

df_cannabis = pd.read_csv(f'{base_path}\\cannabis\\Defendants_Ethnic\\ethnic_cannabis.csv')
df_cocaine  = pd.read_csv(f'{base_path}\\cocaine\\Defendants_Ethnic\\ethnic_cocaine.csv')
df_hashish  = pd.read_csv(f'{base_path}\\hashish\\Defendants_Ethnic\\ethnic_hashish.csv')

type_map = {
    'Before Matching (Naive )': 'Naive',
    'Before Matching (Naive)':  'Naive',
    'After Matching (Standard)': 'Standard',
    'After Matching (my PSM)':   'LLM'
}

def prep(df):
    df['Type'] = df['Type'].str.strip()
    df['Type'] = df['Type'].map(type_map)
    return df

df_cannabis = prep(df_cannabis)
df_cocaine  = prep(df_cocaine)
df_hashish  = prep(df_hashish)

drug_dfs = {
    'Cannabis': df_cannabis,
    'Cocaine':  df_cocaine,
    'Hashish':  df_hashish,
}

def get_row(df, outcome, typ):
    r = df[(df['outcome'] == outcome) & (df['Type'] == typ)]
    return r.iloc[0] if not r.empty else None

# פונקציה לבחירת כמות הכוכביות לפי רמת המובהקות
def get_stars(p_value):
    if p_value < 0.001:
        return '***'
    elif p_value < 0.01:
        return '**'
    elif p_value < 0.05:
        return '*'
    else:
        return ''

# ── Colors ─────────────────────────────────────────────────────────
jewish_color = '#F38B00' 
arab_color   = '#0F61A4' 

col_types  = ['Naive', 'Standard', 'LLM']
col_titles = [
    'Before Matching (Naive)',
    'Standard matching (administrative covariates)',
    'AI-powered matching (our approach)'
]

drug_names = list(drug_dfs.keys())

# ── Figure setup ───────────────────────────────────────────────────
fig, axes = plt.subplots(
    2, 3, figsize=(16, 10),
    gridspec_kw={'height_ratios': [1.1, 1]}
)

def style_ax(ax, is_leftmost):
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color('#999999')
        spine.set_linewidth(1.0)
    
    if is_leftmost:
        ax.tick_params(labelsize=12, left=True, bottom=False, labelleft=True) 
        for label in ax.get_yticklabels():
            label.set_fontweight('bold')
    else:
        ax.tick_params(labelsize=12, left=False, bottom=False, labelleft=False)
        
    ax.yaxis.grid(True, alpha=0.6, color='#CCCCCC', zorder=0, linestyle='--')
    ax.set_axisbelow(True)

# ══════════════════════════════════════════════════════════════════
# ROW 1: Sentencing Bounds
# ══════════════════════════════════════════════════════════════════
width = 0.32
gap = 0.2
offset = (width + gap) / 2

# מיקומים על ציר X
x_low = [0.5, 1.8, 3.1]
x_up  = [4.4, 5.7, 7.0] 
mid_vline = 3.75 

for col, typ in enumerate(col_types):
    ax = axes[0][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    # LOWER BOUND
    for idx, drug in enumerate(drug_names):
        df = drug_dfs[drug]
        r = get_row(df, 'min_punishment', typ)
        if r is None: continue

        j_val, a_val = r['mean_control'], r['mean_treatment']
        p_val = r['p_value']
        
        # חישוב רווחי סמך (95%) מתוך סטיות התקן וגודל המדגם
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 1.96, se_a * 1.96

        pos_j = x_low[idx] - offset
        pos_a = x_low[idx] + offset

        # יצירת העמודות כולל רווחי סמך
        ax.bar(pos_j, j_val, width, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})
        ax.bar(pos_a, a_val, width, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})

        # הוספת טקסט וסוגריים אך ורק אם התוצאה מובהקת
        if p_val < 0.05:
            # הוספת הטקסט מעל לקצה העליון של רווח הסמך
            ax.text(pos_j, j_val + ci_j + 0.4, f'{j_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            ax.text(pos_a, a_val + ci_a + 0.4, f'{a_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            
            stars = get_stars(p_val)
            max_val = max(j_val + ci_j, a_val + ci_a)
            y_base = max_val + 2.0
            y_top  = y_base + 0.5
            
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_val - j_val
            sign = '+' if diff > 0 else ''
            # הדפסת ההפרש עם הכוכביות המותאמות
            ax.text(x_low[idx], y_top + 0.2, f'{sign}{diff:.1f}{stars}', ha='center', va='bottom', fontweight='bold', fontsize=12)

    # UPPER BOUND
    for idx, drug in enumerate(drug_names):
        df = drug_dfs[drug]
        r = get_row(df, 'max_punishment', typ)
        if r is None: continue

        j_val, a_val = r['mean_control'], r['mean_treatment']
        p_val = r['p_value']
        
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 1.96, se_a * 1.96
        
        pos_j = x_up[idx] - offset
        pos_a = x_up[idx] + offset

        ax.bar(pos_j, j_val, width, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})
        ax.bar(pos_a, a_val, width, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})

        if p_val < 0.05:
            ax.text(pos_j, j_val + ci_j + 0.4, f'{j_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            ax.text(pos_a, a_val + ci_a + 0.4, f'{a_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)

            stars = get_stars(p_val)
            max_val = max(j_val + ci_j, a_val + ci_a)
            y_base = max_val + 2.0
            y_top  = y_base + 0.5
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_val - j_val
            sign = '+' if diff > 0 else ''
            ax.text(x_up[idx], y_top + 0.2, f'{sign}{diff:.1f} {stars}', ha='center', va='bottom', fontweight='bold', fontsize=12)

    # קו מפריד
    ax.axvline(x=mid_vline, color='#999999', linestyle=':', linewidth=1.2, zorder=1)

    ax.set_xticks(x_low + x_up)
    ax.set_xticklabels(['Cannabis', 'Cocaine', 'Hashish'] * 2, fontweight='bold', fontsize=12)
    ax.set_xlim(-0.2, 7.7) 
    ax.set_ylim(0, 42) # גבול הותאם כדי שיהיה מקום לרווחי הסמך
    ax.set_yticks(np.arange(0, 41, 5))

    ax.text(1.8, -5.0, 'Lower Bound', ha='center', va='top', fontweight='bold', fontsize=14)
    ax.text(5.7, -5.0, 'Upper Bound', ha='center', va='top', fontweight='bold', fontsize=14)

    ax.set_title(col_titles[col], fontsize=15, fontweight='bold', pad=12)
    if is_leftmost:
        ax.set_ylabel('Sentencing Range Bounds (months)', fontsize=14, fontweight='bold')

# ══════════════════════════════════════════════════════════════════
# ROW 2: Incarceration Rate
# ══════════════════════════════════════════════════════════════════
x_rate = [1.5, 4.0, 6.5] 

width_bot = width * 2.2
offset_bot = (width_bot + gap) / 2

for col, typ in enumerate(col_types):
    ax = axes[1][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    for idx, drug in enumerate(drug_names):
        df = drug_dfs[drug]
        r = get_row(df, 'is_actual_sentence', typ)
        if r is None: continue

        j_pct = r['mean_control']   * 100
        a_pct = r['mean_treatment'] * 100
        p_val = r['p_value']
        
        # חישוב רווחי סמך לאחוזים
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 100 * 1.96, se_a * 100 * 1.96

        pos_j = x_rate[idx] - offset_bot
        pos_a = x_rate[idx] + offset_bot

        ax.bar(pos_j, j_pct, width_bot, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})
        ax.bar(pos_a, a_pct, width_bot, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':1.2})

        if p_val < 0.05:
            ax.text(pos_j, j_pct + ci_j + 1.5, f'{j_pct:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=12)
            ax.text(pos_a, a_pct + ci_a + 1.5, f'{a_pct:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=12)

            stars = get_stars(p_val)
            max_val = max(j_pct + ci_j, a_pct + ci_a)
            y_base = max_val + 8.0
            y_top  = y_base + 2.0
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_pct - j_pct
            sign = '+' if diff > 0 else ''
            ax.text(x_rate[idx], y_top + 1.0, f'{sign}{diff:.1f} pp {stars}', ha='center', va='bottom', fontweight='bold', fontsize=13)

    ax.set_xticks(x_rate)
    ax.set_xticklabels(['Cannabis', 'Cocaine', 'Hashish'], fontweight='bold', fontsize=13)
    ax.set_xlim(0, 8.0)
    ax.set_ylim(0, 125)
    ax.set_yticks([0, 20, 40, 60, 80, 100])

    if is_leftmost:
        ax.set_ylabel('Incarceration Rate (%)', fontsize=14, fontweight='bold')

# ── Shared legend ──────────────────────────────────────────────────
legend_handles = [
    mpatches.Patch(facecolor=jewish_color, edgecolor='black', linewidth=1, label='Jewish'),
    mpatches.Patch(facecolor=arab_color,   edgecolor='black', linewidth=1, label='Arab'),
]
fig.legend(
    handles=legend_handles,
    loc='lower center', ncol=2,
    fontsize=15, 
    prop={'weight': 'bold', 'size': 15},
    frameon=False,
    bbox_to_anchor=(0.5, 0.02)
)

# שליטה במרווחים
plt.subplots_adjust(hspace=0.40, wspace=0.02, bottom=0.1)

# שמירת הקובץ ברזולוציה גבוהה
plt.savefig('plots/figure_1_ethnicity_all_drugs.png', dpi=300, bbox_inches='tight')
print("Saved: figure_1_ethnicity_all_drugs.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# ── הגדרות גופן גלובליות (צר, גבוה ומודגש) ───────────────
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial Narrow', 'Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['font.stretch'] = 'condensed' 
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titleweight'] = 'bold'

# ── Load data ──────────────────────────────────────────────────────
base_path = r'C:\Users\user\OneDrive\טל\projects\justice-tal\bias_weapon\matched_outputs\impute'

df_carrying_only = pd.read_csv(f'{base_path}\\carrying_only\\ethnic\\ethnic_carrying_only.csv')
df_possession_only  = pd.read_csv(f'{base_path}\\possession_only\\ethnic\\ethnic_possession_only.csv')
df_importing_only  = pd.read_csv(f'{base_path}\\importing_only\\ethnic\\ethnic_importing_only.csv')

type_map = {
    'Before Matching (Naive )': 'Naive',
    'Before Matching (Naive)':  'Naive',
    'After Matching (Standard)': 'Standard',
    'After Matching (my PSM)':   'LLM'
}

def prep(df):
    df['Type'] = df['Type'].str.strip()
    df['Type'] = df['Type'].map(type_map)
    return df

df_carrying_only = prep(df_carrying_only)
df_possession_only  = prep(df_possession_only)
df_importing_only  = prep(df_importing_only)

weapon_dfs = {
    'Carrying Only': df_carrying_only,
    'Possession Only': df_possession_only,
    'Importing Only': df_importing_only,
}

def get_row(df, outcome, typ):
    r = df[(df['outcome'] == outcome) & (df['Type'] == typ)]
    return r.iloc[0] if not r.empty else None

# פונקציה לבחירת כמות הכוכביות לפי רמת המובהקות
def get_stars(p_value):
    if p_value < 0.001:
        return '***'
    elif p_value < 0.01:
        return '**'
    elif p_value < 0.05:
        return '*'
    else:
        return ''


from scipy.stats import fisher_exact

def binary_p(r):
    """p-value for a binary outcome. When one group has no variance (e.g. 100% incarcerated)
    the logistic model separates and returns a meaningless p (e.g. 1e-56), so fall back to
    Fisher's exact test on the reconstructed counts."""
    if r['sd_control'] == 0 or r['sd_treatment'] == 0:
        n_t, n_c = int(r['n_treatment']), int(r['n_control'])
        k_t, k_c = int(round(r['mean_treatment'] * n_t)), int(round(r['mean_control'] * n_c))
        return fisher_exact([[k_t, n_t - k_t], [k_c, n_c - k_c]])[1]
    return r['p_value']

# ── Colors ─────────────────────────────────────────────────────────
jewish_color = '#F38B00' 
arab_color   = '#0F61A4' 

col_types  = ['Naive', 'Standard', 'LLM']
col_titles = [
    'Before Matching (Naive)',
    'Standard matching (administrative covariates)',
    'AI-powered matching (our approach)'
]

weapon_names = list(weapon_dfs.keys())

# ── Figure setup ───────────────────────────────────────────────────
fig, axes = plt.subplots(
    2, 3, figsize=(19, 10),
    gridspec_kw={'height_ratios': [1.1, 1]}
)

def style_ax(ax, is_leftmost):
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color('#999999')
        spine.set_linewidth(1.0)
    
    if is_leftmost:
        ax.tick_params(labelsize=12, left=True, bottom=False, labelleft=True) 
        for label in ax.get_yticklabels():
            label.set_fontweight('bold')
    else:
        ax.tick_params(labelsize=12, left=False, bottom=False, labelleft=False)
        
    ax.yaxis.grid(True, alpha=0.6, color='#CCCCCC', zorder=0, linestyle='--')
    ax.set_axisbelow(True)

# ══════════════════════════════════════════════════════════════════
# ROW 1: Sentencing Bounds
# ══════════════════════════════════════════════════════════════════
width = 0.32
gap = 0.2
offset = (width + gap) / 2

# מיקומים על ציר X
x_low = [0.5, 1.8, 3.1]
x_up  = [4.9, 6.2, 7.5]

mid_vline = 4.0

for col, typ in enumerate(col_types):
    ax = axes[0][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    # LOWER BOUND
    for idx, weapon in enumerate(weapon_names):
        df = weapon_dfs[weapon]
        r = get_row(df, 'min_punishment', typ)
        if r is None: continue

        j_val, a_val = r['mean_control'], r['mean_treatment']
        p_val = r['p_value']
        
        # חישוב רווחי סמך (95%) מתוך סטיות התקן וגודל המדגם
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 1.96, se_a * 1.96

        pos_j = x_low[idx] - offset
        pos_a = x_low[idx] + offset

        # יצירת העמודות כולל רווחי סמך
        ax.bar(pos_j, j_val, width, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})
        ax.bar(pos_a, a_val, width, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})

        # הוספת טקסט וסוגריים אך ורק אם התוצאה מובהקת
        if p_val < 0.05:
            # הוספת הטקסט מעל לקצה העליון של רווח הסמך
            ax.text(pos_j, j_val + ci_j + 0.4, f'{j_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            ax.text(pos_a, a_val + ci_a + 0.4, f'{a_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            
            stars = get_stars(p_val)
            max_val = max(j_val + ci_j, a_val + ci_a)
            y_base = max_val + 2.0
            y_top  = y_base + 0.5
            
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_val - j_val
            sign = '+' if diff > 0 else ''
            # הדפסת ההפרש עם הכוכביות המותאמות
            ax.text(x_low[idx], y_top + 0.2, f'{sign}{diff:.1f}{stars}', ha='center', va='bottom', fontweight='bold', fontsize=12)

    # UPPER BOUND
    for idx, weapon in enumerate(weapon_names):
        df = weapon_dfs[weapon]
        r = get_row(df, 'max_punishment', typ)
        if r is None: continue

        j_val, a_val = r['mean_control'], r['mean_treatment']
        p_val = r['p_value']
        
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 1.96, se_a * 1.96
        
        pos_j = x_up[idx] - offset
        pos_a = x_up[idx] + offset

        ax.bar(pos_j, j_val, width, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})
        ax.bar(pos_a, a_val, width, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})

        if p_val < 0.05:
            ax.text(pos_j, j_val + ci_j + 0.4, f'{j_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)
            ax.text(pos_a, a_val + ci_a + 0.4, f'{a_val:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)

            stars = get_stars(p_val)
            max_val = max(j_val + ci_j, a_val + ci_a)
            y_base = max_val + 2.0
            y_top  = y_base + 0.5
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_val - j_val
            sign = '+' if diff > 0 else ''
            ax.text(x_up[idx], y_top + 0.2, f'{sign}{diff:.1f} {stars}', ha='center', va='bottom', fontweight='bold', fontsize=12)

    # קו מפריד
    ax.axvline(x=mid_vline, color='#999999', linestyle=':', linewidth=1.2, zorder=1)

    ax.set_xticks(x_low + x_up)
    ax.set_xticklabels(['Carrying', 'Possession', 'Importing'] * 2, fontweight='bold', fontsize=10)
    ax.set_xlim(-0.2, 8.2)
    ax.set_ylim(0, 130) # גבול הותאם כדי שיהיה מקום לרווחי הסמך
    ax.set_yticks(np.arange(0, 131, 10))

    ax.text(1.8, -14.0, 'Lower Bound', ha='center', va='top', fontweight='bold', fontsize=14)
    ax.text(6.2, -14.0, 'Upper Bound', ha='center', va='top', fontweight='bold', fontsize=14)

    ax.set_title(col_titles[col], fontsize=15, fontweight='bold', pad=12)
    if is_leftmost:
        ax.set_ylabel('Sentencing Range Bounds (months)', fontsize=14, fontweight='bold')

# ══════════════════════════════════════════════════════════════════
# ROW 2: Incarceration Rate
# ══════════════════════════════════════════════════════════════════
x_rate = [1.5, 4.0, 6.5]

width_bot = width * 2.2
offset_bot = (width_bot + gap) / 2

for col, typ in enumerate(col_types):
    ax = axes[1][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    for idx, weapon in enumerate(weapon_names):
        df = weapon_dfs[weapon]
        r = get_row(df, 'is_actual_sentence', typ)
        if r is None: continue

        j_pct = r['mean_control']   * 100
        a_pct = r['mean_treatment'] * 100
        p_val = binary_p(r)
        
        # חישוב רווחי סמך לאחוזים
        n_j, n_a = r['n_control'], r['n_treatment']
        sd_j, sd_a = r['sd_control'], r['sd_treatment']
        
        se_j = sd_j / np.sqrt(n_j) if n_j > 0 else 0
        se_a = sd_a / np.sqrt(n_a) if n_a > 0 else 0
        ci_j, ci_a = se_j * 100 * 1.96, se_a * 100 * 1.96

        pos_j = x_rate[idx] - offset_bot
        pos_a = x_rate[idx] + offset_bot

        ax.bar(pos_j, j_pct, width_bot, yerr=ci_j, capsize=3, color=jewish_color, edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})
        ax.bar(pos_a, a_pct, width_bot, yerr=ci_a, capsize=3, color=arab_color,   edgecolor='black', lw=0.8, zorder=3, error_kw={'elinewidth':0.9})

        if p_val < 0.05:
            ax.text(pos_j, j_pct + ci_j + 1.5, f'{j_pct:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=12)
            ax.text(pos_a, a_pct + ci_a + 1.5, f'{a_pct:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=12)

            stars = get_stars(p_val)
            max_val = max(j_pct + ci_j, a_pct + ci_a)
            y_base = max_val + 8.0
            y_top  = y_base + 2.0
            ax.plot([pos_j, pos_j, pos_a, pos_a], [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_pct - j_pct
            sign = '+' if diff > 0 else ''
            ax.text(x_rate[idx], y_top + 1.0, f'{sign}{diff:.1f} pp {stars}', ha='center', va='bottom', fontweight='bold', fontsize=13)

    ax.set_xticks(x_rate)
    ax.set_xticklabels(['Carrying', 'Possession', 'Importing'], fontweight='bold', fontsize=13)
    ax.set_xlim(0, 8.0)
    ax.set_ylim(0, 125)
    ax.set_yticks([0, 20, 40, 60, 80, 100])

    if is_leftmost:
        ax.set_ylabel('Incarceration Rate (%)', fontsize=14, fontweight='bold')

# ── Shared legend ──────────────────────────────────────────────────
legend_handles = [
    mpatches.Patch(facecolor=jewish_color, edgecolor='black', linewidth=1, label='Jewish'),
    mpatches.Patch(facecolor=arab_color,   edgecolor='black', linewidth=1, label='Arab'),
]
fig.legend(
    handles=legend_handles,
    loc='lower center', ncol=2,
    fontsize=15, 
    prop={'weight': 'bold', 'size': 15},
    frameon=False,
    bbox_to_anchor=(0.5, 0.02)
)

# שליטה במרווחים
plt.subplots_adjust(hspace=0.40, wspace=0.02, bottom=0.1)
# שמירת הקובץ ברזולוציה גבוהה
plt.savefig(rf'plots\figure_3_ethnicity_all_weapon.png', dpi=300, bbox_inches='tight')
print("Saved: figure_ethnicity_all_weapons.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.colors as mcolors
from matplotlib.cm import ScalarMappable
import numpy as np

# =============================================================================
# Plot Settings
# =============================================================================
plt.rcParams.update({
    "font.family": "serif",
    "axes.titlesize": 13,
    "axes.labelsize": 12,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "figure.dpi": 300
})

# =============================================================================
# Load Data
# =============================================================================
weapons = pd.read_csv(
    r"results_CSVs\weapon_district_all_results.csv",
    encoding="utf-8-sig"
)

drugs = pd.read_csv(
    r"results_CSVs\drugs_district_all_results.csv",
    encoding="utf-8-sig"
)

# =============================================================================
# Clean District Names
# =============================================================================
# def clean_district(s):
#     if pd.isna(s):
#         return s
#     return s.replace("district_", "").replace("_", " ").title()

# for df in [drugs, weapons]:
#     df["District"] = df["District"].apply(clean_district)

combined_all = pd.concat([drugs, weapons], ignore_index=True)

# =============================================================================
# Filter to Jerusalem / Center / South only
# =============================================================================
DISTRICT_ORDER = [
    "Jerusalem",
    "Center",
    "South",
]

DISTRICT_LABELS = {
    "Jerusalem": "Jerusalem",
    "Center":    "Center",
    "South":     "South",
}

# DISTRICT_ORDER = [
#     "Tel",
#     "North",
#     "Haifa"
# ]

# DISTRICT_LABELS = {
#     "Tel": "Tel Aviv",
#     "North":     "North",
#     "Haifa":     "Haifa",
# }

# combined_all = combined_all[combined_all["District"].isin(DISTRICT_ORDER)].copy()
# =============================================================================
# Orders
# =============================================================================
OFFENSE_ORDER = [
    "cannabis",
    "cocaine",
    "hashish",
    "possession_only",
    "carrying_only"
]

COL_LABELS = {
    "cannabis": "Cannabis",
    "cocaine": "Cocaine",
    "hashish": "Hashish",
    "possession_only": "Possession",
    "carrying_only": "Carrying",
}

OUTCOMES = {
    "min_punishment":     "Lower bound",
    "max_punishment":     "Upper bound",
    "is_actual_sentence": "Incarceration",
}
OUTCOME_KEYS = list(OUTCOMES.keys())

# =============================================================================
# Build aligned matrices across (District, Outcome) rows
# =============================================================================
def build_matrices(data):
    rows = [(d, o) for d in DISTRICT_ORDER for o in OUTCOME_KEYS]

    p_mat = pd.DataFrame(
        index=pd.MultiIndex.from_tuples(rows, names=["District", "Outcome"]),
        columns=OFFENSE_ORDER, dtype=float
    )
    d_mat = p_mat.copy()

    for (d, o) in rows:
        sub = data[(data["District"] == d) & (data["outcome"] == o)]
        if sub.empty:
            continue
        pcol = "p_bonferroni" if "p_bonferroni" in sub.columns else "p_value"
        piv_p = sub.pivot_table(index="District", columns="Type",
                                values=pcol, aggfunc="first")
        piv_d = sub.pivot_table(index="District", columns="Type",
                                values="diff", aggfunc="first")
        for col in OFFENSE_ORDER:
            if col in piv_p.columns:
                p_mat.loc[(d, o), col] = piv_p.iloc[0][col]
                d_mat.loc[(d, o), col] = piv_d.iloc[0][col]
    return p_mat, d_mat


p_mat, d_mat = build_matrices(combined_all)

# Incarceration differences are proportions -> percentage points, on their own color scale
is_inc = d_mat.index.get_level_values("Outcome") == "is_actual_sentence"
d_mat.loc[is_inc] = d_mat.loc[is_inc] * 100

def sym_norm(vals):
    m = np.nanmax(np.abs(vals))
    m = 1 if (np.isnan(m) or m == 0) else m
    return mcolors.Normalize(vmin=-m, vmax=m)

cmap = plt.get_cmap("RdBu_r")
norm_months = sym_norm(d_mat.loc[~is_inc].values)
norm_pp     = sym_norm(d_mat.loc[is_inc].values)

# =============================================================================
# Draw single combined heatmap (District x Outcome rows, Offense columns)
# =============================================================================
n_rows = len(DISTRICT_ORDER) * len(OUTCOME_KEYS)
n_cols = len(OFFENSE_ORDER)
n_out  = len(OUTCOME_KEYS)

fig, ax = plt.subplots(figsize=(11, 8))

ax.set_xlim(0, n_cols)
ax.set_ylim(0, n_rows)
ax.invert_yaxis()

row_index = 0
for d in DISTRICT_ORDER:
    for o in OUTCOME_KEYS:
        for ci, col in enumerate(OFFENSE_ORDER):
            p_val = p_mat.loc[(d, o), col]
            diff  = d_mat.loc[(d, o), col]

            norm = norm_pp if o == "is_actual_sentence" else norm_months
            color = "#F0F0F0" if pd.isna(diff) else cmap(norm(diff))

            rect = mpatches.Rectangle(
                    (ci, row_index),
                    1, 1,
                    facecolor=color,
                    edgecolor=color,
                    linewidth=1.0,
                    antialiased=False
                )
            ax.add_patch(rect)

            if not pd.isna(diff):
                sig = ""
                if not pd.isna(p_val) and p_val < 0.05:
                    sig = "***" if p_val < 0.001 else ("**" if p_val < 0.01 else "*")
                rgb = mcolors.to_rgb(color)
                luminance = 0.299 * rgb[0] + 0.587 * rgb[1] + 0.114 * rgb[2]
                txt_color = "white" if luminance < 0.45 else "black"
                unit = " pp" if o == "is_actual_sentence" else ""
                ax.text(
                    ci + 0.5,
                    row_index + 0.52,
                    f"{diff:+.1f}{unit}{sig}",
                    ha="center",
                    va="center",
                    fontsize=10.5,
                    fontweight="bold" if sig else "normal",
                    color=txt_color
                )
        row_index += 1

# ----- LEFT Y-AXIS: District labels (spanning their 3 outcome rows) ----------
left_ticks  = [i * n_out + n_out / 2 for i in range(len(DISTRICT_ORDER))]
ax.set_yticks(left_ticks)
ax.set_yticklabels(
    [DISTRICT_LABELS[d] for d in DISTRICT_ORDER],
    fontsize=13,
    fontweight="bold"
)

# Column (offense) labels on bottom
ax.set_xticks([i + 0.5 for i in range(n_cols)])
ax.set_xticklabels(
    [COL_LABELS.get(c, c) for c in OFFENSE_ORDER],
    rotation=35,
    ha="right"
)
ax.tick_params(which="both", bottom=False, left=False, pad=10)

for spine in ax.spines.values():
    spine.set_visible(False)

# Vertical separator between drug and weapon offenses
sep_x = OFFENSE_ORDER.index("possession_only")
ax.axvline(x=sep_x, color="#333333", linewidth=1.4)

# Heavy horizontal lines between districts
for i in range(1, len(DISTRICT_ORDER)):
    ax.axhline(y=i * n_out, color="#333333", linewidth=1.4)

# ----- RIGHT Y-AXIS: Outcome labels (one per row) ----------------------------
ax_right = ax.twinx()
ax_right.set_ylim(ax.get_ylim())
right_ticks = [i + 0.5 for i in range(n_rows)]
right_labels = [OUTCOMES[o] for d in DISTRICT_ORDER for o in OUTCOME_KEYS]
ax_right.set_yticks(right_ticks)
ax_right.set_yticklabels(right_labels, fontsize=10)
ax_right.tick_params(axis="y", length=0, pad=6)
for spine in ax_right.spines.values():
    spine.set_visible(False)
ax.grid(False)
ax_right.grid(False)
# Drug / Weapon group labels above the heatmap
ax.text(
    sep_x / 2,
    -0.4,
    "Drug Offenses",
    ha="center", va="center",
    fontsize=11, fontweight="bold"
)
ax.text(
    sep_x + (n_cols - sep_x) / 2,
    -0.4,
    "Weapon Offenses",
    ha="center", va="center",
    fontsize=11, fontweight="bold"
)

# =============================================================================
# Title
# =============================================================================
# fig.suptitle(
#     "District-Level Sentencing Disparities\n(Jerusalem / Center / South vs. All Others)",
#     fontsize=14,
#     fontweight="bold",
#     y=1.02
# )

# =============================================================================
# Shared Colorbar (pushed further to the right)
# =============================================================================
for (norm_, pos, lab) in [(norm_months, [0.96, 0.53, 0.02, 0.30], "Bounds: difference\n(months)"),
                          (norm_pp,     [0.96, 0.15, 0.02, 0.30], "Incarceration:\ndifference (pp)")]:
    sm = ScalarMappable(norm=norm_, cmap=cmap)
    sm.set_array([])
    cax = fig.add_axes(pos)
    cbar = fig.colorbar(sm, cax=cax)
    cbar.set_label(lab, fontsize=10, fontweight="bold")

# =============================================================================
# Footnote
# =============================================================================
# fig.text(
#     0.5,
#     -0.04,
#     "Red = District Harsher | Blue = District Lighter | *** p<0.001, ** p<0.01, * p<0.05",
#     ha="center",
#     fontsize=10,
#     style="italic"
# )

plt.tight_layout(rect=[0, 0, 0.92, 1])

plt.savefig(
    rf"plots\figure_4_heatmap_district.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial Narrow', 'Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['font.stretch'] = 'condensed'
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titleweight'] = 'bold'

# ── Load & filter ──────────────────────────────────────────────────
df = pd.read_csv(r'results_CSVs\drugs_periods_all_results.csv')
df = df[df['Type'] == 'cannabis'].copy()

def get_row(df, outcome, method, period):
    r = df[
        (df['outcome'] == outcome) &
        (df['Method'] == method) &
        (df['Time'] == period)
    ]
    return r.iloc[0] if not r.empty else None

def get_stars(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    return ''

jewish_color = '#F38B00'
arab_color   = '#0F61A4'

col_types  = ['naive', 'match_custom', 'match_logit']
col_titles = [
    'Before Matching (Naive)',
    'Standard matching (administrative covariates)',
    'AI-powered matching (our approach)'
]
period_names = ['corona', 'war']

# ── Figure ─────────────────────────────────────────────────────────
fig, axes = plt.subplots(
    2, 3, figsize=(15, 10),
    gridspec_kw={'height_ratios': [1.1, 1]}
)

def style_ax(ax, is_leftmost):
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color('#999999')
        spine.set_linewidth(1.0)
    if is_leftmost:
        ax.tick_params(labelsize=12, left=True, bottom=False, labelleft=True)
        for label in ax.get_yticklabels():
            label.set_fontweight('bold')
    else:
        ax.tick_params(labelsize=12, left=False, bottom=False, labelleft=False)
    ax.yaxis.grid(True, alpha=0.6, color='#CCCCCC', zorder=0, linestyle='--')
    ax.set_axisbelow(True)

width  = 0.47
gap    = 0.2
offset = (width + gap) / 2

x_low     = [1.0, 2.8]
x_up      = [5.0, 6.8]
mid_vline = 3.9

# ══════════════════════════════════════════════════════════════════
# ROW 1 – Sentencing Bounds
# ══════════════════════════════════════════════════════════════════
for col, method in enumerate(col_types):
    ax = axes[0][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    for outcome, x_positions in [('min_punishment', x_low), ('max_punishment', x_up)]:
        for idx, period in enumerate(period_names):
            r = get_row(df, outcome, method, period)
            if r is None:
                continue

            j_val, a_val = r['mean_control'], r['mean_treatment']
            p_val        = r['p_value']
            ci_j = 1.96 * r['sd_control']   / np.sqrt(r['n_control'])   if r['n_control']   > 0 else 0
            ci_a = 1.96 * r['sd_treatment']  / np.sqrt(r['n_treatment']) if r['n_treatment'] > 0 else 0

            pos_j = x_positions[idx] - offset
            pos_a = x_positions[idx] + offset

            ax.bar(pos_j, j_val, width, yerr=ci_j, capsize=3,
                   color=jewish_color, edgecolor='black', lw=0.8, zorder=3)
            ax.bar(pos_a, a_val, width, yerr=ci_a, capsize=3,
                   color=arab_color,   edgecolor='black', lw=0.8, zorder=3)

            if p_val < 0.05:
                max_val = max(j_val + ci_j, a_val + ci_a)
                y_base  = max_val + 2
                y_top   = y_base  + 0.5
                ax.plot([pos_j, pos_j, pos_a, pos_a],
                        [y_base, y_top, y_top, y_base], color='black', lw=0.8)
                diff = a_val - j_val
                ax.text(x_positions[idx], y_top + 0.2,
                        f'{("+" if diff > 0 else "")}{diff:.1f}{get_stars(p_val)}',
                        ha='center', fontweight='bold', fontsize=12)

    ax.axvline(x=mid_vline, color='#999999', linestyle=':', linewidth=1.2)
    ax.set_xticks(x_low + x_up)
    ax.set_xticklabels(['COVID-19', 'War', 'COVID-19', 'War'], fontweight='bold', fontsize=12)
    ax.set_xlim(0, 7.8)
    ax.text(1.9, -5, 'Lower Bound', ha='center', fontweight='bold', fontsize=14)
    ax.text(5.9, -5, 'Upper Bound', ha='center', fontweight='bold', fontsize=14)
    ax.set_title(col_titles[col], fontsize=15, fontweight='bold')
    if is_leftmost:
        ax.set_ylabel('Sentencing Range Bounds (months)', fontsize=14, fontweight='bold')

# ══════════════════════════════════════════════════════════════════
# ROW 2 – Incarceration Rate
# ══════════════════════════════════════════════════════════════════
x_rate     = [2.0, 5.5]
width_bot  = width * 2.2
offset_bot = (width_bot + gap) / 2

for col, method in enumerate(col_types):
    ax = axes[1][col]
    is_leftmost = (col == 0)
    style_ax(ax, is_leftmost)

    for idx, period in enumerate(period_names):
        r = get_row(df, 'is_actual_sentence', method, period)
        if r is None:
            continue

        j_pct = r['mean_control']   * 100
        a_pct = r['mean_treatment'] * 100
        p_val = r['p_value']
        ci_j  = 1.96 * r['sd_control']   / np.sqrt(r['n_control'])   * 100 if r['n_control']   > 0 else 0
        ci_a  = 1.96 * r['sd_treatment']  / np.sqrt(r['n_treatment']) * 100 if r['n_treatment'] > 0 else 0

        pos_j = x_rate[idx] - offset_bot
        pos_a = x_rate[idx] + offset_bot

        ax.bar(pos_j, j_pct, width_bot, yerr=ci_j, capsize=3,
               color=jewish_color, edgecolor='black', lw=0.8)
        ax.bar(pos_a, a_pct, width_bot, yerr=ci_a, capsize=3,
               color=arab_color,   edgecolor='black', lw=0.8)

        if p_val < 0.05:
            max_val = max(j_pct + ci_j, a_pct + ci_a)
            y_base  = max_val + 8
            y_top   = y_base  + 2
            ax.plot([pos_j, pos_j, pos_a, pos_a],
                    [y_base, y_top, y_top, y_base], color='black', lw=0.8)
            diff = a_pct - j_pct
            ax.text(x_rate[idx], y_top + 1,
                    f'{("+" if diff > 0 else "")}{diff:.1f} pp {get_stars(p_val)}',
                    ha='center', fontweight='bold', fontsize=13)

    ax.set_xticks(x_rate)
    ax.set_xticklabels(['COVID-19', 'War'], fontweight='bold', fontsize=13)
    ax.set_xlim(0, 7.5)
    ax.set_ylim(0, 125)
    if is_leftmost:
        ax.set_ylabel('Incarceration Rate (%)', fontsize=14, fontweight='bold')

# ── Legend ─────────────────────────────────────────────────────────
fig.legend(
    handles=[
        mpatches.Patch(facecolor=jewish_color, edgecolor='black', label='Pre-period (control)'),
        mpatches.Patch(facecolor=arab_color,   edgecolor='black', label='Post-period (treatment)'),
    ],
    loc='lower center', ncol=2, fontsize=15, frameon=False,
    bbox_to_anchor=(0.5, 0.02)
)

plt.subplots_adjust(hspace=0.40, wspace=0.02, bottom=0.1)
#save
plt.savefig(r'plots\figure_7_barplot_drugs_periods.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# =============================================================================
# STYLE - אופטימלי לוורד
# =============================================================================
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial Narrow', 'Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titleweight'] = 'bold'

# =============================================================================
# LOAD DATA
# =============================================================================
drugs = pd.read_csv(
    r"results_CSVs\weapon_periods_all_results.csv", # שנה לנתיב הרלוונטי במחשב שלך
    encoding="utf-8-sig"
)

def clean_time(s):
    if pd.isna(s):
        return s
    return {"corona": "COVID-19", "war": "War"}.get(s, s.replace("_", " ").title())

# התאמה לעמודת Time במקום District
drugs["Time"] = drugs["Time"].apply(clean_time)

# מציג את כל הנתונים של possession_only (לא מסנן שורות)
drugs = drugs[
    (drugs["Type"].str.lower().str.strip() == "possession_only") &
    (drugs["Method"] == "match_logit")   # primary specification only
].copy()

times = [t for t in ["COVID-19", "War"] if t in set(drugs["Time"].dropna())]

# =============================================================================
# SETTINGS
# =============================================================================
control_color = "#F38B00"  # כתום = Control group
center_color  = "#0F61A4"  # כחול  = Treatment group

outcomes = [
    ("min_punishment",    "Lower"),
    ("max_punishment",    "Upper"),
    ("is_actual_sentence","Incarceration"),
]

width        = 0.32
inner_gap    = 0.8
time_gap     = 0.3

fig, ax = plt.subplots(figsize=(9, 5.5))

x_positions = []
x_labels    = []
current_x   = 0

def get_stars(p):
    if p < 0.001: return "***"
    elif p < 0.01: return "**"
    elif p < 0.05: return "*"
    return ""

# =============================================================================
# חישוב מקסימום מראש
# =============================================================================
max_y_value = 0
for time_val in times:
    time_df = drugs[drugs["Time"] == time_val]
    for outcome, _ in outcomes:
        row = time_df[time_df["outcome"] == outcome]
        if row.empty:
            continue
        row = row.iloc[0]
        j_val = row["mean_control"] * (100 if outcome == "is_actual_sentence" else 1)
        a_val = row["mean_treatment"] * (100 if outcome == "is_actual_sentence" else 1)
        n_j, n_a = row["n_control"], row["n_treatment"]
        sd_j, sd_a = row["sd_control"], row["sd_treatment"]
        ci_j = 1.96 * sd_j / np.sqrt(n_j) * (100 if outcome == "is_actual_sentence" else 1) if n_j > 0 else 0
        ci_a = 1.96 * sd_a / np.sqrt(n_a) * (100 if outcome == "is_actual_sentence" else 1) if n_a > 0 else 0
        max_y_value = max(max_y_value, j_val + ci_j, a_val + ci_a)

# =============================================================================
# PLOT
# =============================================================================
for time_val in times:
    time_df = drugs[drugs["Time"] == time_val]
    time_start_x = current_x
    plotted_any = False

    for outcome, short_name in outcomes:
        row = time_df[time_df["outcome"] == outcome]
        if row.empty:
            continue

        row = row.iloc[0]
        plotted_any = True

        scale  = 100 if outcome == "is_actual_sentence" else 1
        j_val  = row["mean_control"]  * scale   # Control
        a_val  = row["mean_treatment"] * scale  # Treatment 
        p_val  = row["p_value"]
        n_j, n_a   = row["n_control"], row["n_treatment"]
        sd_j, sd_a = row["sd_control"], row["sd_treatment"]
        ci_j = 1.96 * sd_j / np.sqrt(n_j) * scale if n_j > 0 else 0
        ci_a = 1.96 * sd_a / np.sqrt(n_a) * scale if n_a > 0 else 0

        bar_gap = 0.03
        pos_j = current_x - width / 2 - bar_gap / 2   # Treatment = Left
        pos_a = current_x + width / 2 + bar_gap / 2   # Control = Right

        ax.bar(pos_j, j_val, width, color=control_color, edgecolor="black",
               linewidth=0.7, yerr=ci_j, capsize=2.5,
               error_kw={'linewidth': 0.8}, zorder=3)

        ax.bar(pos_a, a_val, width, color=center_color, edgecolor="black",
               linewidth=0.7, yerr=ci_a, capsize=2.5,
               error_kw={'linewidth': 0.8}, zorder=3)

        stars = get_stars(p_val)
        
        # --- כאן בוצע השינוי: מציג את ההבדל והקו רק אם יש כוכביות ---
        if stars:
            offset  = max_y_value * 0.025
            max_val = max(j_val + ci_j, a_val + ci_a)
            y_base  = max_val + offset
            y_top   = y_base  + offset * 0.5

            ax.plot([pos_j, pos_j, pos_a, pos_a],
                    [y_base, y_top, y_top, y_base],
                    color="black", lw=0.7)

            diff = a_val - j_val   # Treatment minus Control
            sign = "+" if diff > 0 else ""
            suffix = " pp" if outcome == "is_actual_sentence" else ""

            ax.text(current_x, y_top + offset * 0.3,
                    f"{sign}{diff:.1f}{suffix}{stars}",
                    ha="center", va="bottom", fontsize=8.5, fontweight="bold")
        # -------------------------------------------------------------

        x_positions.append(current_x)
        x_labels.append(short_name)
        current_x += inner_gap

    if plotted_any:
        time_end_x    = current_x - inner_gap
        time_center_x = (time_start_x + time_end_x) / 2

        ax.text(time_center_x, -max_y_value * 0.09, time_val,
                ha="center", va="top", fontsize=9, fontweight="bold")

        current_x += time_gap

# =============================================================================
# STYLE
# =============================================================================
ax.set_xticks(x_positions)
ax.set_xticklabels(x_labels, fontsize=8.5, fontweight="bold")

ax.grid(axis="y", linestyle="--", alpha=0.35, linewidth=0.6)
ax.set_axisbelow(True)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

ax.set_ylabel("Sentence (months) / Incarceration (%)", fontsize=11, fontweight="bold")
if max_y_value > 0:
    ax.set_ylim(0, max_y_value * 1.1)
ax.tick_params(axis='y', labelsize=9)

legend_handles = [
    mpatches.Patch(facecolor=center_color,  edgecolor="black", linewidth=0.7, label="Post-period (treatment)"),
    mpatches.Patch(facecolor=control_color, edgecolor="black", linewidth=0.7, label="Pre-period (control)"),
]

ax.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.12),
    ncol=2,
    frameon=False,
    fontsize=10
)

plt.subplots_adjust(bottom=0.18, left=0.07, right=0.98, top=0.92)
plt.savefig(
    rf"plots\figure_8_barplot_weapon_periods.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# =============================================================================
# STYLE
# =============================================================================
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial Narrow", "Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.weight"] = "bold"
plt.rcParams["axes.labelweight"] = "bold"
plt.rcParams["axes.titleweight"] = "bold"

treatment_color = "#0F61A4"
control_color   = "#F38B00"

outcomes = [
    ("min_punishment",    "Lower"),
    ("max_punishment",    "Upper"),
    ("is_actual_sentence","Incarceration"),
]

width      = 0.32
inner_gap  = 0.8   # between outcomes within one ethnic group
ethnic_gap = 0.45  # between Arab and Jewish within one drug/weapon type
type_gap   = 0.9   # between drug/weapon types


def get_stars(p):
    if pd.isna(p):
        return ""
    if p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    return ""


def make_ethnic_chart(df, types_list, type_labels, ethnic_groups, save_path=None):
    fig, ax = plt.subplots(figsize=(14, 5.5))

    x_positions = []
    x_labels    = []
    current_x   = 0

    # ── max_y_value pass ──────────────────────────────────────────────────────
    max_y_value = 0
    for drug_type in types_list:
        for ethnic in ethnic_groups:
            sub = df[(df["Type"] == drug_type) & (df["Time"] == ethnic)]
            for outcome, _ in outcomes:
                row = sub[sub["outcome"] == outcome]
                if row.empty:
                    continue
                r     = row.iloc[0]
                scale = 100 if outcome == "is_actual_sentence" else 1
                t_val = r["mean_treatment"] * scale
                c_val = r["mean_control"]   * scale
                ci_t  = 1.96 * r["sd_treatment"] / np.sqrt(r["n_treatment"]) * scale if r["n_treatment"] > 0 else 0
                ci_c  = 1.96 * r["sd_control"]   / np.sqrt(r["n_control"])   * scale if r["n_control"]   > 0 else 0
                max_y_value = max(max_y_value, t_val + ci_t, c_val + ci_c)

    # ── plot pass ─────────────────────────────────────────────────────────────
    type_sections = []   # (center_x, label)

    for ti, drug_type in enumerate(types_list):
        type_x_positions = []
        ethnic_sections  = []

        for ei, ethnic in enumerate(ethnic_groups):
            sub            = df[(df["Type"] == drug_type) & (df["Time"] == ethnic)]
            ethnic_start_x = current_x
            plotted_any    = False

            for outcome, short_name in outcomes:
                row = sub[sub["outcome"] == outcome]
                if row.empty:
                    continue
                r          = row.iloc[0]
                plotted_any = True

                scale = 100 if outcome == "is_actual_sentence" else 1
                t_val = r["mean_treatment"] * scale
                c_val = r["mean_control"]   * scale
                p_val = r["p_value"]
                ci_t  = 1.96 * r["sd_treatment"] / np.sqrt(r["n_treatment"]) * scale if r["n_treatment"] > 0 else 0
                ci_c  = 1.96 * r["sd_control"]   / np.sqrt(r["n_control"])   * scale if r["n_control"]   > 0 else 0

                bar_gap = 0.03
                pos_c = current_x - width / 2 - bar_gap / 2
                pos_t = current_x + width / 2 + bar_gap / 2

                ax.bar(
                    pos_t, t_val, width,
                    color=treatment_color, edgecolor="black", linewidth=0.7,
                    yerr=ci_t if ci_t > 0 else None, capsize=2.5,
                    error_kw={"linewidth": 0.8}, zorder=3
                )
                ax.bar(
                    pos_c, c_val, width,
                    color=control_color, edgecolor="black", linewidth=0.7,
                    yerr=ci_c if ci_c > 0 else None, capsize=2.5,
                    error_kw={"linewidth": 0.8}, zorder=3
                )

                offset  = max_y_value * 0.025
                max_val = max(t_val + ci_t, c_val + ci_c)
                y_base  = max_val + offset
                y_top   = y_base  + offset * 0.5

                if p_val < 0.05:
                    ax.plot(
                        [pos_t, pos_t, pos_c, pos_c],
                        [y_base, y_top, y_top, y_base],
                        color="black", lw=0.7
                    )

                    diff   = t_val - c_val
                    sign   = "+" if diff > 0 else ""
                    suffix = "%" if outcome == "is_actual_sentence" else ""
                    ax.text(
                        current_x, y_top + offset * 0.3,
                        f"{sign}{diff:.1f}{suffix}{get_stars(p_val)}",
                        ha="center", va="bottom", fontsize=8.5, fontweight="bold"
                    )

                type_x_positions.append(current_x)
                x_positions.append(current_x)
                x_labels.append(short_name)
                current_x += inner_gap

            if plotted_any:
                ethnic_end_x    = current_x - inner_gap
                ethnic_center_x = (ethnic_start_x + ethnic_end_x) / 2
                ethnic_sections.append((ethnic_center_x, ethnic))
                if ei < len(ethnic_groups) - 1:
                    current_x += ethnic_gap

        # ethnic labels
        for ecx, elabel in ethnic_sections:
            ax.text(
                ecx, -max_y_value * 0.09, elabel,
                ha="center", va="top", fontsize=9, fontweight="bold", clip_on=False
            )

        if type_x_positions:
            type_center_x = (type_x_positions[0] + type_x_positions[-1]) / 2
            type_sections.append((type_center_x, type_labels[ti]))

        if ti < len(types_list) - 1:
            current_x += type_gap

    # type super-labels
    for tcx, tlabel in type_sections:
        ax.text(
            tcx, -max_y_value * 0.20, tlabel,
            ha="center", va="top", fontsize=10, fontweight="bold",
            color="#111111", clip_on=False
        )

    # ── axes style ────────────────────────────────────────────────────────────
    ax.set_xticks(x_positions)
    ax.set_xticklabels(x_labels, fontsize=8.5, fontweight="bold")

    ax.grid(axis="y", linestyle="--", alpha=0.35, linewidth=0.6)
    ax.set_axisbelow(True)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_linewidth(0.8)
    ax.spines["bottom"].set_linewidth(0.8)

    ax.set_ylabel("Sentence (months) / Incarceration (%)", fontsize=11, fontweight="bold")
    ax.set_ylim(0, max_y_value * 1.22)
    ax.tick_params(axis="y", labelsize=9)

    legend_handles = [
        mpatches.Patch(
            facecolor=treatment_color, edgecolor="black", linewidth=0.7,
            label="Treatment group: war period"
        ),
        mpatches.Patch(
            facecolor=control_color, edgecolor="black", linewidth=0.7,
            label="Control group: pre war period"
        ),
    ]
    ax.legend(
        handles=legend_handles,
        loc="upper center", bbox_to_anchor=(0.5, -0.30),
        ncol=2, frameon=False, fontsize=10
    )

    plt.subplots_adjust(bottom=0.33, left=0.08, right=0.98, top=0.90)

    plt.savefig(rf"plots\figure_9_barplot_drugs_periods_ethnic.png", dpi=300, bbox_inches="tight", facecolor="white")

    plt.show()

print('helpers loaded')


# =============================================================================
# DRUGS CHART  (cannabis + cocaine  ×  Arab / Jewish)
# =============================================================================
drugs = pd.read_csv(
    r"results_CSVs\drugs_all_results_times_ethnic.csv",
    encoding="utf-8-sig"
)

make_ethnic_chart(
    df           = drugs,
    types_list   = ["cannabis", "cocaine"],
    type_labels  = ["Cannabis", "Cocaine"],
    ethnic_groups= ["Arab", "Jewish"],
)
```

# power analsys

```python
import math
import numpy as np
import pandas as pd
from statsmodels.stats.power import TTestIndPower, NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

# =============================================================================
# UNIFIED POWER ANALYSIS - every matched (PSM) comparison from results_CSVs
#   one-sided ('larger'), alpha = 0.05, target power = 0.80
#   reports required N per group + power actually achieved at the observed N
#
# EFFECT SIZE (a priori / SESOI approach, per advisor):
#   Continuous outcomes (min/max punishment, in MONTHS):
#       es = SESOI_MONTHS / pooled_std        (smallest difference worth detecting)
#       pooled_std = sqrt(((n1-1)*sd1^2 + (n2-1)*sd2^2) / (n1+n2-2))
#   Binary outcome (is_actual_sentence):
#       es = SESOI_H (fixed Cohen's h, e.g. 76% -> 91% incarceration at the cannabis baseline)
# =============================================================================
BASE         = r"results_CSVs"
ALPHA        = 0.05
POWER        = 0.80
ALT          = "larger"   # one-sided
SESOI_MONTHS = 6         # smallest meaningful sentence difference (months)
SESOI_H      = 0.40      # smallest meaningful incarceration difference (Cohen's h)

# (readable name, file, extra grouping column)
ANALYSES = [
    ("Drugs - Ethnic (Jewish vs Arab)",   "drugs_ethnic_all_results.csv",         None),
    ("Drugs - District",                  "drugs_district_all_results.csv",       "District"),
    ("Drugs - Period (post vs pre)",      "drugs_periods_all_results.csv",        "Time"),
    ("Drugs - Period x Ethnic",           "drugs_all_results_times_ethnic.csv",   "Time"),
    ("Weapons - Ethnic (Jewish vs Arab)", "weapon_ethnic_all_results.csv",        None),
    ("Weapons - District",                "weapon_district_all_results.csv",      "District"),
    ("Weapons - Period (post vs pre)",    "weapon_periods_all_results.csv",        "Time"),
    ("Weapons - Period x Ethnic",         "weapons_all_results_times_ethnic.csv", "Time"),
]

t_engine = TTestIndPower()
z_engine = NormalIndPower()


def safe_float(x):
    try:
        return float(x)
    except (TypeError, ValueError):
        return np.nan


def pooled_std(sd1, sd2, n1, n2):
    """Pooled within-group standard deviation (Cohen's d denominator)."""
    if any(pd.isna(v) for v in (sd1, sd2, n1, n2)) or (n1 + n2 - 2) <= 0:
        return np.nan
    return math.sqrt(((n1 - 1) * sd1 ** 2 + (n2 - 1) * sd2 ** 2) / (n1 + n2 - 2))


rows = []
for analysis_name, fname, group_col in ANALYSES:
    df = pd.read_csv(f"{BASE}\\{fname}", encoding="utf-8-sig")

    for _, r in df.iterrows():
        outcome   = r["outcome"]
        is_binary = (outcome == "is_actual_sentence")   # robust binary flag
        n_t       = safe_float(r["n_treatment"])
        n_c       = safe_float(r["n_control"])
        group     = str(r[group_col]) if group_col else "-"

        # ---- effect size ----
        if is_binary:
            es       = SESOI_H
            engine   = z_engine
            var_type = f"Binary (Z, SESOI h={SESOI_H})"
        else:
            # SESOI: standardize a meaningful 2-month difference by the pooled SD
            s_p      = pooled_std(safe_float(r["sd_treatment"]), safe_float(r["sd_control"]), n_t, n_c)
            es       = (SESOI_MONTHS / s_p) if (np.isfinite(s_p) and s_p > 0) else np.nan
            engine   = t_engine
            var_type = f"Continuous (t, SESOI={SESOI_MONTHS}mo)"

        # ---- required N per group & achieved power ----
        req_n, achieved = np.nan, np.nan
        if np.isfinite(es) and es > 0 and n_t >= 2 and n_c >= 2:
            try:
                req_n = math.ceil(engine.solve_power(
                    effect_size=es, alpha=ALPHA, power=POWER, alternative=ALT))
            except Exception:
                req_n = np.nan
            try:
                achieved = engine.power(
                    effect_size=es, nobs1=n_t, alpha=ALPHA,
                    ratio=n_c / n_t, alternative=ALT)
            except Exception:
                achieved = np.nan

        p_val = safe_float(r["p_value"])
        rows.append({
            "Analysis":      analysis_name,
            "Type":          r["Type"],
            "Group":         group,
            "Method":        r["Method"],
            "Outcome":       outcome,
            "VarType":       var_type,
            "EffectSize":    round(es, 3) if np.isfinite(es) else np.nan,
            "ReqN_perGroup": req_n,
            "ActualN":       int(n_t) if np.isfinite(n_t) else np.nan,
            "AchievedPower": round(achieved, 3) if np.isfinite(achieved) else np.nan,
            "Underpowered":  (achieved < POWER) if np.isfinite(achieved) else None,
            "p_value":       round(p_val, 4) if np.isfinite(p_val) else np.nan,
        })

power_df = (
    pd.DataFrame(rows)
    .sort_values(["Analysis", "Type", "Group", "Method", "Outcome"])
    .reset_index(drop=True)
)

pd.set_option("display.max_rows", None)
pd.set_option("display.width", 200)

print("=" * 48 + " POWER ANALYSIS - ALL COMPARISONS " + "=" * 48)
print(power_df.to_string(index=False))

n_eval  = int(power_df["Underpowered"].notna().sum())
n_under = int((power_df["Underpowered"] == True).sum())
print("\n" + "-" * 70)
print(f"Under-powered (achieved power < {POWER}): {n_under} of {n_eval} evaluable comparisons")

out_path = f"{BASE}\\power_analysis_all.csv"
power_df.to_csv(out_path, index=False, encoding="utf-8-sig")
print(f"Saved -> {out_path}")
```

```python
import os
import glob
import pandas as pd

ROOT   = r"C:\Users\user\OneDrive\טל\projects\justice-tal"
SM_DIR = r"results_CSVs\SM"
os.makedirs(SM_DIR, exist_ok=True)

DOMAINS = {
    "Drugs":   ("bias",        ["cannabis", "cocaine", "hashish"]),
    "Weapons": ("bias_weapon", ["possession_only", "carrying_only", "importing_only"]),
}
KINDS = {
    "impute":      "Primary (RF imputation, ambiguous lower bound = 3)",
    "impute_0":    "Ambiguous lower bound = 0",
    "impute_6":    "Ambiguous lower bound = 6",
    "no_impute":   "Listwise deletion",
    "no_personal": "Without age / marital status / criminal record",
    "unfiltered":  "Unfiltered sample (incarceration only)",
}
KEEP_COLS = ["outcome", "mean_treatment", "mean_control", "diff", "odds_ratio",
             "ci_low", "ci_high", "d_cohens", "p_value", "p_bonferroni",
             "rosenbaum_gamma", "e_value_point", "eval_ci",
             "n_treatment_unique", "n_control_unique", "OR", "n_pairs"]

def read_results(path, **labels):
    if not os.path.exists(path):
        return None
    df = pd.read_csv(path, encoding="utf-8-sig")
    df = df[[c for c in KEEP_COLS if c in df.columns]]
    return df.assign(**labels)

def analyses_in(offense_dir):
    # (analysis, comparison, folder) for every PSM analysis inside one offense folder
    out = [("Ethnicity", "Arab vs Jewish", "Defendants_Ethnic"),
           ("Period", "COVID (post vs pre)", "corona"),
           ("Period", "War (post vs pre)", "war")]
    for sector in ["Arab", "Jewish"]:
        out.append(("Period within sector", f"War (post vs pre), {sector}",
                    os.path.join("time_by_sector", sector)))
    for folder in sorted(glob.glob(os.path.join(offense_dir, "district_*_vs_All_Others"))):
        district = os.path.basename(folder)[len("district_"):-len("_vs_All_Others")].replace("_", " ")
        out.append(("District (one vs rest)", f"{district} vs rest", os.path.basename(folder)))
    return out

def signif_p(row):
    # Bonferroni-corrected p for district comparisons, nominal p otherwise
    return row["p_bonferroni"] if row["Analysis"].startswith("District") else row["p_value"]

def fmt_cell(row):
    if pd.isna(row["diff"]):
        return ""
    p = row["p_used"]
    stars = "***" if p < .001 else "**" if p < .01 else "*" if p < .05 else ""
    n = row.get("n_treatment_unique")
    n_txt = f", n={int(n)}" if pd.notna(n) else ""
    return f"{row['diff']:+.2f}{stars} (p={p:.3f}{n_txt})"
```

## SM – selection into the filtered sample (Section 3.3.3)
Is exclusion for a missing sentencing range related to defendant ethnicity or case severity? Logistic regression of `excluded` on ethnicity, offense type, severity, criminal record, court and district, within the unfiltered sample of each corpus.

```python
import numpy as np
import statsmodels.formula.api as smf

SEVERITY = {
    "Drugs":   "np.log1p(total_amount)",
    "Weapons": "np.log1p(bullets) + np.log1p(weapon_count)",
}
WEAPON_COUNT_COLS = ["handgun", "submachine_gun", "assault_rifle", "sniper_rifle", "hunting_rifle",
                     "improvised_firearm", "grenade", "explosive_charge", "molotov", "anti_tank_missile",
                     "ammunition_only", "weapon_parts", "unidentified_firearm"]

rows, rates = [], []
for domain, (track, offenses) in DOMAINS.items():
    frames = []
    for offense in offenses:
        data_dir = os.path.join(ROOT, track, "data")
        unfiltered = pd.read_csv(os.path.join(data_dir, "unfiltered", f"{offense}.csv"), encoding="utf-8-sig")
        filtered   = pd.read_csv(os.path.join(data_dir, "impute", f"{offense}.csv"), encoding="utf-8-sig")
        unfiltered["excluded"] = (~unfiltered["original_case"].isin(filtered["original_case"])).astype(int)
        frames.append(unfiltered.assign(Offense=offense))
    df = pd.concat(frames, ignore_index=True)
    df = df[df["ethnic"].isin(["Arab", "Jewish"])].copy()
    df["arab"] = (df["ethnic"] == "Arab").astype(int)
    if domain == "Weapons":
        # -1 (count unknown) is treated as 0, as in the matching code (log1p(pmax(x, 0)))
        counts = df[[c for c in WEAPON_COUNT_COLS if c in df.columns]].apply(pd.to_numeric, errors="coerce")
        df["weapon_count"] = counts.clip(lower=0).sum(axis=1)
        df["bullets"] = pd.to_numeric(df["bullets"], errors="coerce").fillna(0).clip(lower=0)

    rates.append(df.groupby("ethnic")["excluded"].agg(["mean", "sum", "count"])
                 .rename(columns={"mean": "excluded_share", "sum": "n_excluded", "count": "n"})
                 .reset_index().assign(Domain=domain))

    formula = (f"excluded ~ arab + C(Offense) + {SEVERITY[domain]} + C(Criminal_record)"
               " + C(Court) + C(district)")
    fit = smf.logit(formula, data=df).fit(disp=False)
    ci = fit.conf_int()
    for term in ["arab"] + [t for t in fit.params.index if t.startswith("np.log1p")]:
        rows.append({"Domain": domain, "term": term, "OR": np.exp(fit.params[term]),
                     "OR_ci_low": np.exp(ci.loc[term, 0]), "OR_ci_high": np.exp(ci.loc[term, 1]),
                     "p_value": fit.pvalues[term], "n": int(fit.nobs), "n_excluded": int(df["excluded"].sum())})

selection = pd.DataFrame(rows)
exclusion_rates = pd.concat(rates, ignore_index=True)[["Domain", "ethnic", "n", "n_excluded", "excluded_share"]]
selection.to_csv(os.path.join(SM_DIR, "selection_into_filtered_sample.csv"), index=False, encoding="utf-8-sig")
exclusion_rates.to_csv(os.path.join(SM_DIR, "exclusion_rates_by_ethnicity.csv"), index=False, encoding="utf-8-sig")
print(exclusion_rates.to_string(index=False), "\n")
print(selection.to_string(index=False))
```

## SM – descriptive statistics (Section 3.1.5)
Final analytical samples by offense type and defendant ethnicity. Personal characteristics are shown as observed (before imputation), with the share of unknown values.

```python
import numpy as np

UNKNOWN = "לא ידוע"
PERSONAL_COLS = ["age", "marital_status", "Criminal_record"]
LABELS = {
    "רווק": "Single", "נשוי": "Married", "גרוש": "Divorced", "אלמן": "Widowed",
    "ללא עבר פלילי": "None", "עבר פלילי קל": "Light", "עבר פלילי כבד": "Serious",
    "לא נגרם": "None", "קל": "Light", "גבוה": "Severe", "לא צוין": "Not stated",
    UNKNOWN: "Unknown",
}
WEAPON_TYPES = ["handgun", "submachine_gun", "assault_rifle", "sniper_rifle", "hunting_rifle",
                "improvised_firearm", "grenade", "explosive_charge", "molotov", "anti_tank_missile",
                "ammunition_only", "weapon_parts", "unidentified_firearm"]

def load_observed(track, offense):
    # Primary analytical sample, with the imputed personal columns restored to their observed values
    data_dir = os.path.join(ROOT, track, "data")
    df = pd.read_csv(os.path.join(data_dir, "impute", f"{offense}.csv"), encoding="utf-8-sig")
    observed = (pd.read_csv(os.path.join(data_dir, "features_extractions.csv"), encoding="utf-8-sig")
                .set_index("original_case")[PERSONAL_COLS])
    df[PERSONAL_COLS] = observed.loc[df["original_case"]].values
    return df

def pct(mask):
    return f"{100 * mask.mean():.1f}"

def mean_sd(x):
    x = pd.to_numeric(x, errors="coerce").dropna()
    return f"{x.mean():.1f} ({x.std():.1f})" if len(x) else ""

def median_iqr(x):
    x = pd.to_numeric(x, errors="coerce").dropna()
    return f"{x.median():.1f} [{x.quantile(.25):.1f}–{x.quantile(.75):.1f}]" if len(x) else ""

def describe(df, domain):
    date = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce")
    age = pd.to_numeric(df["age"], errors="coerce")
    rows = [("N", str(len(df))),
            ("Age at offense, mean (SD)", mean_sd(age)),
            ("Age unknown, %", pct(age.isna()))]
    marital = ["רווק", "נשוי", "גרוש"] + (["אלמן"] if domain == "Weapons" else []) + [UNKNOWN]
    for v in marital:
        rows.append((f"Marital status: {LABELS[v]}, %", pct(df["marital_status"] == v)))
    for v in ["ללא עבר פלילי", "עבר פלילי קל", "עבר פלילי כבד", UNKNOWN]:
        rows.append((f"Criminal record: {LABELS[v]}, %", pct(df["Criminal_record"] == v)))
    rows += [("Guilty plea, %", pct(pd.to_numeric(df["confession"], errors="coerce") == 1)),
             ("Expressed regret, %", pct(pd.to_numeric(df["regret"], errors="coerce") == 1)),
             ("District court (vs. Magistrate), %", pct(df["Court"] == "מחוזי")),
             ("Arab judge (panel), %", pct(df["judge_ethnic"] == "Arab"))]
    for d in ["Center", "Haifa", "Jerusalem", "North", "South", "Tel Aviv"]:
        rows.append((f"District: {d}, %", pct(df["district"] == d)))
    rows += [("Adjudicated from March 2020 (COVID), %", pct(date >= "2020-03-01")),
             ("Adjudicated from October 2023 (war), %", pct(date >= "2023-10-07"))]
    if domain == "Drugs":
        rows += [("Drug quantity, grams, median [IQR]", median_iqr(df["total_amount"])),
                 ("Personal use, %", pct(df["self_use"] == 1)),
                 ("Laboratory / growing, %", pct(df["laboratory"] == 1)),
                 ("Sold to undercover agent, %", pct(df["sold_to_agent"] == 1)),
                 ("Production offense (production_offence), %", pct(df["production_offence"] == 1)),
                 ("Possession / use offense (use_offence), %", pct(df["use_offence"] == 1)),
                 ("Trafficking / importation offense (importation_offence), %", pct(df["importation_offence"] == 1))]
    else:
        for w in ["handgun", "assault_rifle", "submachine_gun", "improvised_firearm", "grenade", "explosive_charge"]:
            rows.append((f"Weapon type: {w.replace('_', ' ')}, %", pct(pd.to_numeric(df[w], errors="coerce") > 0)))
        bullets = pd.to_numeric(df["bullets"], errors="coerce")
        rows += [("Weapons seized, median [IQR]", median_iqr(df[WEAPON_TYPES].apply(pd.to_numeric, errors="coerce").clip(lower=0).sum(axis=1))),
                 ("Bullets, median [IQR] (count stated)", median_iqr(bullets[bullets >= 0])),
                 ("Bullets present, count not stated, %", pct(bullets == -1)),
                 ("Premeditation, %", pct(pd.to_numeric(df["premeditation"], errors="coerce") == 1)),
                 ("Escape attempt, %", pct(pd.to_numeric(df["escape_attempt"], errors="coerce") == 1)),
                 ("Sold to undercover officer, %", pct(pd.to_numeric(df["undercover_sale"], errors="coerce") == 1)),
                 ("Independent (non-weapon) offenses, %", pct(pd.to_numeric(df["has_independent_offenses"], errors="coerce") == 1))]
        for v in ["לא נגרם", "קל", "גבוה"]:
            rows.append((f"Actual damage: {LABELS[v]}, %", pct(df["actual_damage"] == v)))
    prison = pd.to_numeric(df["punishment_months"], errors="coerce")
    rows += [("Incarceration, %", pct(pd.to_numeric(df["is_actual_sentence"], errors="coerce") == 1)),
             ("Sentencing range lower bound, months, mean (SD)", mean_sd(df["min_punishment"])),
             ("Sentencing range upper bound, months, mean (SD)", mean_sd(df["max_punishment"])),
             ("Prison term if incarcerated, months, mean (SD)", mean_sd(prison[prison > 0]))]
    return pd.Series(dict(rows))

for domain, (track, offenses) in DOMAINS.items():
    columns = {}
    for offense in offenses:
        df = load_observed(track, offense)
        for group in ["Arab", "Jewish", "All"]:
            sub = df if group == "All" else df[df["ethnic"] == group]
            columns[(offense, group)] = describe(sub, domain)
    table = pd.DataFrame(columns)
    table.columns = pd.MultiIndex.from_tuples(table.columns, names=["Offense", "Ethnicity"])
    table.to_csv(os.path.join(SM_DIR, f"descriptive_statistics_{domain.lower()}.csv"), encoding="utf-8-sig")
    print(f"=== {domain} ===")
    print(table.to_string(), "\n")
```

## SM – LLM extraction validation (Section 3.2.3)
Accuracy against the human-coded ground truth (125 cases per corpus) and by defendant ethnicity, with 95% bootstrap CIs. Inter-coder agreement (κ / ICC) is reported separately.

```python
VALIDATION = {
    # domain: (folder, file suffix, minority group label)
    "Drugs":   (os.path.join(ROOT, "bias", "llm_results"), "Defendants_Ethnic", "Not Jewish"),
    "Weapons": (os.path.join(ROOT, "bias_weapon", "llm_results"), "ethnic", "Arab"),
}

def with_ci(df, metric):
    return df.apply(lambda r: f"{r[metric]:.3f} [{r[metric + '_ci_lower']:.3f}, {r[metric + '_ci_upper']:.3f}]", axis=1)

def ci_overlap(a, b, metric):
    return (a[metric + "_ci_lower"] <= b[metric + "_ci_upper"]) & (b[metric + "_ci_lower"] <= a[metric + "_ci_upper"])

tables, summary = [], []
for domain, (folder, suffix, minority) in VALIDATION.items():
    for kind, metric, extra, prefix in [("Categorical", "f1", ["accuracy"], "categorical"),
                                        ("Continuous", "mae", ["rmse"], "numeric")]:
        df = pd.read_csv(os.path.join(folder, f"{prefix}_metrics_{suffix}.csv"))
        by = {g: d.set_index("variable") for g, d in df.groupby("group")}
        all_, jew, mino = by["All"], by["Jewish"], by[minority]
        out = pd.DataFrame({
            "Domain": domain, "Type": kind, "Metric": metric.upper(),
            "n": all_["n"],
            "All": with_ci(all_, metric),
            "Jewish": with_ci(jew.loc[all_.index], metric),
            "Arab / non-Jewish": with_ci(mino.loc[all_.index], metric),
            "CIs overlap": ci_overlap(jew.loc[all_.index], mino.loc[all_.index], metric).map({True: "yes", False: "NO"}),
        })
        for e in extra:
            out[f"{e} (All)"] = all_[e].round(3)
        tables.append(out.reset_index())
        summary.append({"Domain": domain, "Type": kind, "variables": len(all_),
                        f"mean {metric.upper()}": round(all_[metric].mean(), 3),
                        "min": f"{all_[metric].min():.3f} ({all_[metric].idxmin()})",
                        "CIs not overlapping": int((out['CIs overlap'] == 'NO').sum())})

validation = pd.concat(tables, ignore_index=True).rename(columns={"variable": "Variable"})
validation.to_csv(os.path.join(SM_DIR, "llm_validation_full.csv"), index=False, encoding="utf-8-sig")
print(pd.DataFrame(summary).to_string(index=False), "\n")
print(validation.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

PLOT_DIR = os.path.join("plots", "SM")
os.makedirs(PLOT_DIR, exist_ok=True)

COVARIATE_LABELS = {
    "distance": "Propensity score", "age": "Age", "year": "Year", "total_amount": "Drug quantity (log)",
    "Courtמחוזי": "District court", "ethnicArab": "Defendant Arab", "judge_ethnicArab": "Judge Arab",
    "Criminal_recordללא עבר פלילי": "Criminal record: none", "Criminal_recordעבר פלילי קל": "Criminal record: light",
    "Criminal_recordעבר פלילי כבד": "Criminal record: serious",
    "marital_statusרווק": "Single", "marital_statusנשוי": "Married", "marital_statusגרוש": "Divorced",
    "marital_statusאלמן": "Widowed",
    "actual_damageלא נגרם": "Damage: none", "actual_damageקל": "Damage: light", "actual_damageגבוה": "Damage: severe",
    "actual_damageלא צוין": "Damage: not stated",
    "confession1": "Guilty plea", "confession": "Guilty plea", "regret1": "Expressed regret", "regret": "Expressed regret",
    "self_use1": "Personal use", "laboratory1": "Laboratory / growing", "sold_to_agent1": "Sold to agent",
    "production_offence1": "Production offense", "use_offence1": "Possession / use offense",
    "importation_offence1": "Trafficking / importation offense",
    "premeditation1": "Premeditation", "premeditation": "Premeditation",
    "escape_attempt1": "Escape attempt", "escape_attempt": "Escape attempt",
    "undercover_sale1": "Undercover sale", "undercover_sale": "Undercover sale",
    "has_independent_offenses1": "Independent offenses", "has_independent_offenses": "Independent offenses",
    "magazines": "Magazines (log)", "bullets": "Bullets (log)",
}
DROP_REDUNDANT = {"Courtשלום", "ethnicJewish", "judge_ethnicJewish"}

def clean_balance(path):
    b = pd.read_csv(path, encoding="utf-8-sig")
    # binary 0/1 dummies carry the same |SMD| twice — keep the "1" level only
    b = b[~b["Covariate"].str.endswith("0") & ~b["Covariate"].isin(DROP_REDUNDANT)].copy()
    b["label"] = [COVARIATE_LABELS.get(c, (c[:-1] if c.endswith("1") else c).replace("_", " ").replace("district", "District: "))
                  for c in b["Covariate"]]
    b["before"] = b["Std. Mean Diff._before"].abs()
    b["after"] = b["Std. Mean Diff._after"].abs()
    return b.sort_values("before")

def analysis_panels(track, offense):
    # (family, panel title, folder) for the primary analyses of one offense type
    base = os.path.join(ROOT, track, "matched_outputs", "impute", offense)
    ethnic_folder = "ethnic" if track == "bias_weapon" else "Defendants_Ethnic"
    panels = [("ethnicity", "Arab vs Jewish", ethnic_folder)]
    for folder in sorted(glob.glob(os.path.join(base, "district_*_vs_All_Others"))):
        name = os.path.basename(folder)[len("district_"):-len("_vs_All_Others")].replace("_", " ")
        panels.append(("district", f"{name} vs rest", os.path.basename(folder)))
    panels += [("period", "COVID (post vs pre)", "corona"), ("period", "War (post vs pre)", "war"),
               ("sector", "War, Arab defendants", os.path.join("time_by_sector", "Arab")),
               ("sector", "War, Jewish defendants", os.path.join("time_by_sector", "Jewish"))]
    return [(f, t, p) for f, t, p in panels
            if os.path.exists(os.path.join(base, p, "match_logit", "balance_table.csv"))], base

def grid(n, ncol):
    nrow = int(np.ceil(n / ncol))
    return nrow, min(n, ncol)

def love_figure(domain, track, offense, family, panels, base):
    nrow, ncol = grid(len(panels), 3)
    balances = [clean_balance(os.path.join(base, p, "match_logit", "balance_table.csv")) for _, _, p in panels]
    height = max(len(b) for b in balances) * 0.22 + 1.2
    fig, axes = plt.subplots(nrow, ncol, figsize=(4.6 * ncol, height * nrow), squeeze=False)
    for ax, (_, title, _), b in zip(axes.flat, panels, balances):
        y = np.arange(len(b))
        ax.hlines(y, b["after"], b["before"], color="#cccccc", lw=1, zorder=1)
        ax.scatter(b["before"], y, facecolors="white", edgecolors="#555555", s=22, label="Before matching", zorder=2)
        ax.scatter(b["after"], y, color="#1f5f99", s=22, label="After matching", zorder=3)
        ax.axvline(0.25, ls="--", color="#b03a2e", lw=1)
        ax.set_yticks(y)
        ax.set_yticklabels(b["label"], fontsize=7)
        ax.set_title(title, fontsize=9)
        ax.set_xlabel("|ASMD|", fontsize=8)
        ax.tick_params(axis="x", labelsize=7)
    for ax in axes.flat[len(panels):]:
        ax.axis("off")
    handles, labels = axes.flat[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="lower center", ncol=2, frameon=False, fontsize=8)
    fig.suptitle(f"Covariate balance — {domain}, {offense.replace('_', ' ')}, {family}", fontsize=10)
    fig.tight_layout(rect=(0, 0.03, 1, 0.97))
    path = os.path.join(PLOT_DIR, f"love_{domain.lower()}_{offense}_{family}.png")
    fig.savefig(path, dpi=200)
    plt.close(fig)
    return path

def overlap_figure(domain, track, offense, panels):
    base = os.path.join(ROOT, track, "matched_outputs", "ps_overlap", offense)
    found = []
    for family, title, folder in panels:
        folder = "Defendants_Ethnic" if family == "ethnicity" else folder
        path = os.path.join(base, folder, "match_logit", "ps_all.csv")
        if os.path.exists(path):
            found.append((title, pd.read_csv(path)))
    if not found:
        return None
    nrow, ncol = grid(len(found), 4)
    fig, axes = plt.subplots(nrow, ncol, figsize=(3.4 * ncol, 2.6 * nrow), squeeze=False)
    colors = {0: "#e0a030", 1: "#1f5f99"}   # control = gold, treatment = blue (as in ps_boxplot.png)
    for ax, (title, ps) in zip(axes.flat, found):
        # Before matching: all cases; after matching: matched cases only
        groups = [ps.loc[ps["treated"] == 0, "distance"], ps.loc[ps["treated"] == 1, "distance"],
                  ps.loc[(ps["treated"] == 0) & (ps["matched"] == 1), "distance"],
                  ps.loc[(ps["treated"] == 1) & (ps["matched"] == 1), "distance"]]
        box = ax.boxplot(groups, positions=[0.8, 1.2, 1.8, 2.2], widths=0.32, patch_artist=True,
                         medianprops={"color": "black"}, flierprops={"markersize": 2})
        for patch, treated in zip(box["boxes"], [0, 1, 0, 1]):
            patch.set_facecolor(colors[treated])
        ax.set_xticks([1, 2])
        ax.set_xticklabels(["Before\n(all cases)", "After\n(matched)"], fontsize=7)
        ax.set_ylim(-0.02, 1.02)
        n_t = int((ps["treated"] == 1).sum())
        n_m = int(((ps["treated"] == 1) & (ps["matched"] == 1)).sum())
        ax.set_title(f"{title}\nmatched {n_m} of {n_t} treated", fontsize=8)
        ax.tick_params(axis="y", labelsize=7)
        ax.set_ylabel("Propensity score", fontsize=7)
    for ax in axes.flat[len(found):]:
        ax.axis("off")
    handles = [Patch(facecolor=colors[1], label="Treatment"), Patch(facecolor=colors[0], label="Control")]
    fig.legend(handles=handles, loc="lower center", ncol=2, frameon=False, fontsize=8)
    fig.suptitle(f"Propensity score distributions before and after matching — {domain}, {offense.replace('_', ' ')}", fontsize=10)
    fig.tight_layout(rect=(0, 0.05, 1, 0.95))
    path = os.path.join(PLOT_DIR, f"overlap_{domain.lower()}_{offense}.png")
    fig.savefig(path, dpi=200)
    plt.close(fig)
    return path

saved, balance_rows = [], []
for domain, (track, offenses) in DOMAINS.items():
    for offense in offenses:
        panels, base = analysis_panels(track, offense)
        for family, title, folder in panels:
            b = clean_balance(os.path.join(base, folder, "match_logit", "balance_table.csv"))
            b = b[b["Covariate"] != "distance"]
            above = b[b["after"] > 0.25]
            balance_rows.append({"Domain": domain, "Offense": offense, "Analysis": family, "Comparison": title,
                                 "max |ASMD| before": round(b["before"].max(), 3),
                                 "max |ASMD| after": round(b["after"].max(), 3),
                                 "covariates > 0.25 after": len(above),
                                 "which": ", ".join(f"{l} ({v:.2f})" for l, v in zip(above["label"], above["after"]))})
        for family in ["ethnicity", "district", "period", "sector"]:
            fam_panels = [p for p in panels if p[0] == family]
            if fam_panels:
                saved.append(love_figure(domain, track, offense, family, fam_panels, base))
        overlap = overlap_figure(domain, track, offense, panels)
        if overlap:
            saved.append(overlap)
balance_summary = pd.DataFrame(balance_rows)
balance_summary.to_csv(os.path.join(SM_DIR, "balance_summary.csv"), index=False, encoding="utf-8-sig")
print(balance_summary.drop(columns="which").to_string(index=False), "\n")
print(f"Saved {len(saved)} figures to {PLOT_DIR}")
for p in saved:
    print("  ", os.path.basename(p))
```

## SM – ethnic disparities under three levels of covariate adjustment (Figures 1 and 3)
Naïve (before matching), standard matching (administrative covariates) and AI-powered matching, for every offense type. For incarceration the difference is reported in percentage points, computed from the group rates.

```python
LEVELS = {
    "Naive":      os.path.join("match_logit", "results_before_matching.csv"),
    "Standard":   os.path.join("match_custom", "results.csv"),
    "AI-powered": os.path.join("match_logit", "results.csv"),
}
ETHNIC_FOLDER = {"Drugs": "Defendants_Ethnic", "Weapons": "ethnic"}

rows = []
for domain, (track, offenses) in DOMAINS.items():
    for offense in offenses:
        base = os.path.join(ROOT, track, "matched_outputs", "impute", offense, ETHNIC_FOLDER[domain])
        for level, rel in LEVELS.items():
            path = os.path.join(base, rel)
            if not os.path.exists(path):
                print(f"Missing: {domain} / {offense} / {level}")
                continue
            df = pd.read_csv(path, encoding="utf-8-sig")
            binary = df["outcome"] == "is_actual_sentence"
            # binary `diff` is a log-odds in some files — report percentage points from the rates instead
            df.loc[binary, "diff"] = 100 * (df.loc[binary, "mean_treatment"] - df.loc[binary, "mean_control"])
            df.loc[binary, ["ci_low", "ci_high"]] = np.nan
            n_t = df["n_treatment_unique"] if "n_treatment_unique" in df else df["n_treatment"]
            n_c = df["n_control_unique"] if "n_control_unique" in df else df["n_control"]
            rows.append(df.assign(Domain=domain, Offense=offense, Level=level, n_arab=n_t, n_jewish=n_c)
                        [["Domain", "Offense", "Level", "outcome", "mean_treatment", "mean_control", "diff",
                          "ci_low", "ci_high", "d_cohens", "odds_ratio", "or_ci_low", "or_ci_high", "p_value",
                          "n_arab", "n_jewish"]])

three_levels = pd.concat(rows, ignore_index=True)
three_levels.to_csv(os.path.join(SM_DIR, "ethnicity_three_levels.csv"), index=False, encoding="utf-8-sig")
print(three_levels.groupby(["Domain", "Level"])["outcome"].count())
```

## SM – treatment effects from the doubly robust outcome models
The `treated` coefficient of the covariate-adjusted model estimated on each matched sample (primary specification). B is in months for the bounds and a log-odds for incarceration.

```python
rows = []
for domain, (track, offenses) in DOMAINS.items():
    for offense in offenses:
        offense_dir = os.path.join(ROOT, track, "matched_outputs", "impute", offense)
        for analysis, comparison, folder in analyses_in(offense_dir):
            if folder == "Defendants_Ethnic":
                folder = ETHNIC_FOLDER[domain]
            path = os.path.join(offense_dir, folder, "match_logit", "full_model_coefficients.csv")
            if not os.path.exists(path):
                continue
            coefs = pd.read_csv(path, encoding="utf-8-sig")
            treated = coefs[coefs["term"] == "treated"]
            rows.append(treated.assign(Domain=domain, Offense=offense, Analysis=analysis, Comparison=comparison)
                        [["Domain", "Offense", "Analysis", "Comparison", "outcome", "B", "SE", "CI_low", "CI_high",
                          "p_value", "R_squared"]])

adjusted = pd.concat(rows, ignore_index=True)
adjusted.to_csv(os.path.join(SM_DIR, "adjusted_treatment_effects.csv"), index=False, encoding="utf-8-sig")
print(adjusted.groupby(["Domain", "Analysis"])["outcome"].count())
```

## SM – distribution of the quantity covariates (log transformation, Section 3.2.2)
Skewness of the quantity covariates before and after the transformation used in the propensity score models: `log(kg + 0.001)` for drug quantity (src.ipynb, drugs) and `log1p(max(x, 0))` for bullets, magazines and weapons seized (src.ipynb, weapons; an unstated bullet count, coded -1, becomes 0).

```python
from scipy.stats import skew

WEAPON_TYPES = ["handgun", "submachine_gun", "assault_rifle", "sniper_rifle", "hunting_rifle",
                "improvised_firearm", "grenade", "explosive_charge", "molotov", "anti_tank_missile",
                "ammunition_only", "weapon_parts", "unidentified_firearm"]

series = []   # (domain, offense, variable, raw values, transformed values)
for offense in DOMAINS["Drugs"][1]:
    df = pd.read_csv(os.path.join(ROOT, "bias", "data", "impute", f"{offense}.csv"), encoding="utf-8-sig")
    grams = pd.to_numeric(df["total_amount"], errors="coerce").dropna()
    series.append(("Drugs", offense, "Drug quantity (grams)", grams, np.log(grams / 1000 + 0.001)))
for offense in DOMAINS["Weapons"][1]:
    df = pd.read_csv(os.path.join(ROOT, "bias_weapon", "data", "impute", f"{offense}.csv"), encoding="utf-8-sig")
    counts = df[[c for c in WEAPON_TYPES if c in df.columns]].apply(pd.to_numeric, errors="coerce").clip(lower=0)
    for label, raw in [("Bullets", pd.to_numeric(df["bullets"], errors="coerce")),
                       ("Magazines", pd.to_numeric(df["magazines"], errors="coerce")),
                       ("Weapons seized", counts.sum(axis=1))]:
        raw = raw.dropna()
        series.append(("Weapons", offense, label, raw, np.log1p(raw.clip(lower=0))))

rows = []
for domain, offense, variable, raw, transformed in series:
    stated = raw[raw >= 0]   # median / IQR over stated counts (bullets: -1 = count not stated)
    rows.append({"Domain": domain, "Offense": offense, "Variable": variable, "n": len(raw),
                 "Median [IQR]": f"{stated.median():.1f} [{stated.quantile(.25):.1f}–{stated.quantile(.75):.1f}]",
                 "Max": raw.max(), "Unstated (-1), %": round(100 * (raw == -1).mean(), 1) if variable == "Bullets" else np.nan,
                 "Skewness, raw": round(skew(raw.clip(lower=0)), 2), "Skewness, transformed": round(skew(transformed), 2)})
skew_table = pd.DataFrame(rows)
skew_table.to_csv(os.path.join(SM_DIR, "quantity_skewness.csv"), index=False, encoding="utf-8-sig")
print(skew_table.to_string(index=False))

# Figure: raw vs transformed distribution for each quantity variable (importing omitted: n = 71)
shown = [s for s in series if s[1] != "importing_only"]
fig, axes = plt.subplots(len(shown), 2, figsize=(8.5, 1.9 * len(shown)), squeeze=False)
for (domain, offense, variable, raw, transformed), (ax_raw, ax_log) in zip(shown, axes):
    name = f"{offense.replace('_only', '').capitalize()}: {variable}"
    ax_raw.hist(raw.clip(lower=0), bins=30, color="#8c8c8c")
    ax_log.hist(transformed, bins=30, color="#1f5f99")
    ax_raw.set_title(f"{name} — raw (skewness {skew(raw.clip(lower=0)):.1f})", fontsize=8)
    ax_log.set_title(f"{name} — transformed (skewness {skew(transformed):.1f})", fontsize=8)
    for ax in (ax_raw, ax_log):
        ax.tick_params(labelsize=7)
        ax.set_ylabel("Cases", fontsize=7)
fig.tight_layout()
skew_path = os.path.join(PLOT_DIR, "quantity_distributions.png")
fig.savefig(skew_path, dpi=200)
plt.close(fig)
print("Saved", skew_path)
```
